Every year, American high school students take SATs, which are standardized tests intended to measure literacy, numeracy, and writing skills. There are three sections - reading, math, and writing, each with a **maximum score of 800 points**. These tests are extremely important for students and colleges, as they play a pivotal role in the admissions process.

Analyzing the performance of schools is important for a variety of stakeholders, including policy and education professionals, researchers, government, and even parents considering which school their children should attend. 

You have been provided with a dataset called `schools.csv`, which is previewed below.

You have been tasked with answering three key questions about New York City (NYC) public school SAT performance.

```python
# Re-run this cell 
import pandas as pd

# Read in the data
schools = pd.read_csv("schools.csv")

# Start coding here...
mask=schools["average_math"]>=(0.8*800)
best_math_schools=schools[mask][["school_name","average_math"]].sort_values("average_math",ascending=False)
combined_schools=schools.copy()
combined_schools["total_SAT"]=combined_schools["average_math"]+combined_schools["average_reading"]+combined_schools["average_writing"]
top_10_schools=combined_schools.sort_values("total_SAT",ascending=False).iloc[0:10,[0,7]]
analysis_borough=combined_schools.groupby("borough")["total_SAT"].agg(["std","mean"]).reset_index()
mask_max=analysis_borough["std"]==analysis_borough["std"].max()
largest_std_dev=analysis_borough[mask_max]
largest_borough=str(largest_std_dev.iloc[0,0])
largest_std_dev["num_schools"]=(schools["borough"]==largest_borough).sum()
largest_std_dev=largest_std_dev.rename(columns={"mean":"average_SAT","std":"std_SAT"})
largest_std_dev=largest_std_dev.round(2)


```
